```python
import os
import csv
```

```python
folder_path = "../arquivos"
def generate_sql_from_csvs_in_a_folder(folder_path):
    #checkar se o caminho da pasta espificada existe existe
    sql_schema = ""
    if not os.path.isdir(folder_path):
        print(f"Error: A pasta '{folder_path}' não existe. Corrija o caminho da pasta.")
        return
    #iterarar na lista de arquivos presentes na pasta
    for file_name in os.listdir(folder_path):
        if file_name.endswith(".csv"):
            file_path = os.path.join(folder_path, file_name)

            # Gerar tabelas a partir do nome dos arquivos
            table_name = os.path.splitext(file_name)[0]

            try:
                with open(file_path, mode="r", encoding="utf-8-sig") as f:
                    reader = csv.reader(f)
                    headers = next(reader)  # Read the first row for headers

                # Pegar o nome das colunas e definir o tipo com TEXT
                columns = []
                for header in headers:
                    if header:
                        columns.append(f"    {header} TEXT")

                # Criat a tabela com as colunas
                if columns:
                    sql_schema += f"CREATE TABLE {table_name} (\n"
                    sql_schema += ",\n".join(columns)
                    sql_schema += "\n);\n"


            except Exception as e:
                print(f"-- Erro ao ler o arquivo '{file_name}': {e}\n")


    sql_folder_path = "../questão-2/"
    os.makedirs(sql_folder_path, exist_ok=True)
        
    sql_file_name = "schema.sql"
    with open(os.path.join(sql_folder_path, sql_file_name), "w", encoding="utf-8") as file:
        file.write(sql_schema)        
    print(f"SQL schema: '{sql_schema}'")


generate_sql_from_csvs_in_a_folder(folder_path)
#todo: correctly infer the type of all tables columns
```
